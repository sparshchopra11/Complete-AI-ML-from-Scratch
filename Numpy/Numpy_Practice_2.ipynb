```python
v = np.array([3, 4])
v_mean = v.mean()
print(v_mean)
v_std = np.std(v)
print(v_std)
normalized_v = (v - v_mean)/v_std
print(normalized_v)
print(normalized_v.mean())
print(np.std(normalized_v))
```
