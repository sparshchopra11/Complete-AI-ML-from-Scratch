```python
import numpy as np
import pandas as pd
```

```python
netflix_dataset = pd.read_csv("netflix_titles.csv")
```

```python
print(f"The number of rows are :->  {netflix_dataset.shape[0]} ")
print(f"The number of columns are :->  {netflix_dataset.shape[1]}")
print(f"The names of the columns are :-> {netflix_dataset.columns} ")
print(f"The data types are :->  {netflix_dataset.dtypes}")
netflix_dataset.head()
netflix_dataset.tail()
print(netflix_dataset.isnull().sum())
print(netflix_dataset.nunique())
print(netflix_dataset.duplicated().sum())
print(netflix_dataset.describe())
```

```python
netflix_dataset_copy =  netflix_dataset.copy()
netflix_dataset_copy.isnull().sum()
```

```python
netflix_dataset_copy['director'] = netflix_dataset['director'].fillna("Unknown")
netflix_dataset_copy['cast'] = netflix_dataset['cast'].fillna("Unknown")
netflix_dataset_copy['country'] = netflix_dataset['country'].fillna("Unknown")
netflix_dataset_copy['date_added'] = netflix_dataset['date_added'].fillna(0)
netflix_dataset_copy['duration'] = netflix_dataset['duration'].fillna(0)
netflix_dataset_copy['rating'] = netflix_dataset['rating'].fillna(0)
netflix_dataset_copy.isnull().sum()
```

```python
# title
# director
# country
# rating
# listed_in
netflix_dataset_copy['type'] = netflix_dataset_copy['type'].str.lower()
netflix_dataset_copy['title'] = netflix_dataset_copy['title'].str.lower()
netflix_dataset_copy['director'] = netflix_dataset_copy['director'].str.lower()
netflix_dataset_copy['country'] = netflix_dataset_copy['country'].str.lower()
netflix_dataset_copy['rating'] = netflix_dataset_copy['rating'].str.upper()
netflix_dataset_copy['listed_in'] = netflix_dataset_copy['listed_in'].str.lower()
netflix_dataset_copy
```

```python
tv_show_titles = netflix_dataset_copy.query("type == 'tv show'")['title']
#print(tv_show_titles.shape)
title_length_tv = len(tv_show_titles.sum())/2676
movie_show_titles = netflix_dataset_copy.query("type == 'movie'")['title']
movie_show_titles.shape
title_length_movie = len(movie_show_titles.sum())/6131

print(f"The mean of the title length of the tv show is :->  {title_length_tv}")
print(f"The mean of the title length of the movie is :->  {title_length_movie}")

# we can do the same for the description too 
tv_show_description = netflix_dataset_copy.query("type == 'tv show'")['description']
tv_show_description_len = len(tv_show_description.sum())/2676
movie_description = netflix_dataset_copy.query("type == 'movie'")['description']
movie_description_len = len(movie_description.sum())/6131
print(f"The mean of the decription length of the tv show is :->  {tv_show_description_len}")
print(f"The mean of the decription length of the movie is :->  {movie_description_len}")

# Hence the movie are seen to have longer description and title length 
```

```python
movies = netflix_dataset_copy.query("type == 'movie'")
movies_rows = movies.shape[0]
tv_shows = netflix_dataset_copy.query("type == 'tv show'")
tv_show_rows = tv_shows.shape[0]
print(f"The Movies percentage on netflix are :->  {(movies_rows/8807)*100}")
print(f"The Tv Shows percentage on netflix are :->  {(tv_show_rows/8807)*100}")
# Hence the Netflix is more of the Movies
```

```python
# what is the oldest title 
netflix_dataset_copy['date_added'] = pd.to_datetime(netflix_dataset_copy['date_added'] , errors='coerce')
netflix_dataset_copy['date_added'].max()
netflix_dataset_copy['date_added'].min()

netflix_dataset_copy['date_added'].value_counts()
# hence 2020-01-01 has the most titles

date = pd.to_datetime("2015-01-01")
netflix_dataset_copy.query("date_added > @date")['date_added'].shape
# total are 8655

date1 = pd.to_datetime("2000-01-01")
netflix_dataset_copy.query("date_added < @date1")['date_added'].shape
# no title before 2000
```
